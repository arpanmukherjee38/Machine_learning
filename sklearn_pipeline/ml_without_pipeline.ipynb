```python
import pandas as pd

# Fix for Sex
sex_df = pd.DataFrame(test_input[:, 1].reshape(1, 1), columns=['Sex'])
test_input_sex = ohe_sex.transform(sex_df)

# Fix for Embarked (Assuming 'Embarked' is the 7th item, index 6)
embarked_df = pd.DataFrame(test_input[:, 6].reshape(1, 1), columns=['Embarked'])
test_input_embarked = ohe_embarked.transform(embarked_df)
```

```python
# 1. Extract the Age value (index 2) and reshape it to a 2D array (1 row, 1 column)
test_input_age = test_input[:, 2].reshape(1, 1)

# Note: If you used a SimpleImputer or StandardScaler on Age during training,
# you need to transform it here instead of just extracting it. For example:
# test_input_age = age_imputer.transform(test_input[:, 2].reshape(1, 1))

# 2. Now your concatenation will work perfectly!
test_input_transformed = np.concatenate((
    test_input[:, [0,3,4,5]], # Pclass, SibSp, Parch, Fare (Already a 2D slice)
    test_input_age,           # Age (Now defined!)
    test_input_sex,           # Sex (One-hot encoded)
    test_input_embarked       # Embarked (One-hot encoded)
), axis=1)
```
